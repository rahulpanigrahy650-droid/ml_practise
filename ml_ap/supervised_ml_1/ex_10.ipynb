```python
from sklearn.svm import SVR
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from sklearn import datasets
```

```python
df = datasets.load_diabetes(as_frame=True).frame
```

```python
df.head()
```

```python
x = df.drop("target",axis =1)
y = df["target"]
```

```python
x.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
y_scaler =  StandardScaler()
Y_train = y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
Y_test = y_scaler.fit_transform(y_test.values.reshape(-1,1)).ravel()
```

```python
model = SVR()
model.fit(x_train,Y_train)
y_test_pred = model.predict(x_test)
#y_train_pred = model.predict(x_train)
```

```python
print(f"r2 = {r2_score(Y_test,y_test_pred)}")
#print(f"r2 = {r2_score(Y_test,y_train_pred)}")
```

```python
model_1 = SVR(kernel='linear')
model.fit(x_train,Y_train)
y_test_pred = model.predict(x_test)
```

```python
model_2 = SVR(kernel="poly")
model.fit(x_train,Y_train)
y_test_pred = model.predict(x_test)
```

```python
model_3 = SVR(kernel="sigmoid")
model.fit(x_train,Y_train)
y_test_pred = model.predict(x_test)
```

```python
data = {SVR():"model",
        SVR(kernel="linear") : "model_1",
        SVR(kernel="poly"):"model_2",
        SVR(kernel="sigmoid"):"model_3"
        }

for model,name in data.items():
    model.fit(x_train,Y_train)
    y_test_pred = model.predict(x_test)
    print(f"r2 = {r2_score(Y_test,y_test_pred)} for model = {model}")
```
